```sql
select * from Emps
where salary is null
```

```sql
select * from ord
where discount is not null
```

```sql
select count(manager_id) from Emps 
where manager_id is null
```

```sql
select *,ifnull(salary,0)as salary from emps
```

```sql
select ifnull(stock,0) as stock from products
```

```sql
select coalesce(salary,bonus,0) as earnings from emps
```

```sql
select coalesce(price,1000) as product_price from products
```

```sql
select nullif(salary,0) from emps
```

```sql
select ifnull(salary,0) + ifnull(bonus,0) as earninings from emps
```

```sql
select abs(ifnull(amount,0)-ifnull(discount,0)) from ord
```
