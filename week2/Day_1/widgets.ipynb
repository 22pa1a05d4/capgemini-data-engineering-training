```python
dbutils.widgets.combobox(
    name="Fruits",
    choices=['apple','banana','grape','pineapple'],
    defaultValue='Apple',
    label='fruit_list'
)
```

```python
dbutils.widgets.dropdown(
    name="Fruits_db",
    defaultValue='apple',
    choices=['apple','banana','grape','pineapple'],
    label='fruit_list'
)
```

```python
dbutils.widgets.multiselect(
    name="Fruits_ms",
    defaultValue='apple',
    choices=['apple','banana','grape','pineapple'],
    label='fruit_multiselects'
)
```

```python
dbutils.widgets.text(
    name="Fruits_txt",
    defaultValue='apple',
    label='fruit_text'
)
```
