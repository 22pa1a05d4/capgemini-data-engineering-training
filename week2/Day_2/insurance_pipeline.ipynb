```python
spark.sql("""
    with temp as(
        select customer_id,sum(claim_amount) as claim,sum(premium) as total
        from df
        group by customer_id
    )
select customer_id,claim,total,
case when total =0 then null
else claim/total
end as risk_score
from temp
""").display()
```

```python
spark.sql("""
            with temp as(
                select city,customer_id,sum(coalesce(claim_amount,0)) as claim,sum(coalesce(premium,0)) as total,
                case 
                when sum(coalesce(premium,0)) = 0 then null
                else round(sum(coalesce(claim_amount,0))*1.0/sum(coalesce(premium,0)))
                end as risk_score
                from df
                group by city,customer_id
            ),
            ranked as(
                select *,row_number() over(Partition by city order By risk_score desc) as rn
                from temp
            )
            select *
            from ranked
            where rn<=3

""" ).display()

```

```python
spark.sql('''
          with agg as(
              select customer_id,customer_name,month(claim_date) as month_claim,
              sum(coalesce(claim_amount,0)) as total
              from df
              group by customer_id,customer_name,month(claim_date)
              order by month_claim
          ),
          lagged as(
              select *,
              lag(total) over(partition by customer_id order by month_claim) as lagged
              from agg
          )
          select *
          from lagged
          where total>lagged

''').display()
```

```python
spark.sql('''
          with tem as(
              select agent_id,agent_name,city,sum(premium) as ttl
              from df
              group by agent_id,agent_name,city
              
          ),
          ranked as(
              select *,rank() over( partition by agent_id order by ttl desc) as rnk
              from tem
          )
          select *
         from ranked
          ''').display()
```

```python
# Rank customers based on risk score
spark.sql("""
          with temp as(
              select customer_id,sum(coalesce(claim_amount,0)) as claim,sum(coalesce(premium,0)) as total,
              case 
              when sum(coalesce(premium,0)) =0 then null
              else sum(coalesce(claim_amount,0))*1.0/sum(coalesce(premium,0))
              end as risk_score
              from df
              group by customer_id
          )
          select *,
          rank() over(partition by customer_id  order by risk_score desc)
                    from temp 
          
          """).display()
```
