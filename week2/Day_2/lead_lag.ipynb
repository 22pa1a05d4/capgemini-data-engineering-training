```python
   from pyspark.sql import SparkSession
from pyspark.sql.functions import *
spark = SparkSession.builder.appName("Spark DataFrames").getOrCreate()
df=spark.createDataFrame( [(1,"2024-01-01",100,"A"),(2,"2024-01-03",200,"B"),(3,"2024-01-05",300,"C"),(4,"2024-01-02",500,"A"),(5,"2024-01-06",700,"B"),(6,"2024-01-08",900,"C")],["user_id","date","amount","product"])
df=df.withColumn(
    "date",
    to_date('date','yyyy-MM-dd')
)
df.display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import *
w=Window.orderBy("date")
previous_amount=df.withColumn(
    'previous_amount',
    lag('amount',1).over(w)
)

previous_amount.display()
```

```python

from pyspark.sql.window import Window
from pyspark.sql.functions import *
w=Window.orderBy("date") 
next_amount=df.withColumn(
    'next_amount',
    lead('amount',1).over(w)
)
next_amount.display()
```

```python

from pyspark.sql.window import Window
from pyspark.sql.functions import *
w=Window.orderBy("date") 
first_amount=df.withColumn(
    'first_amount',
    first("amount").over(w)
)
first_amount.display()
```

```python
previous_amount=df.withColumn(
    'previous_amount',
    lag("amount",1).over(w)
)
filtered=previous_amount.filter(col("amount")>col("previous_amount"))
filtered.display()
```

```python
difference=df.withColumn(
    'difference',
    col("amount")-lead("amount",1).over(w)
)
difference.display()
```
