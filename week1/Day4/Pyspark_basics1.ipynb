```python
from pyspark.sql.functions import col,when
df = spark.read.csv("/Volumes/workspace/default/sample/C1.csv",header=True,inferSchema=True)

```

```python
df.show()
```

```python
df=df.withColumn(
    "Sales",
    when(col("Sales")=="blank","0").otherwise(col("Sales"))
    ).withColumn(
        "Category",
        when(col("Category")=="Blank","Unknown").otherwise(col("Category"))
        ).withColumn(
            "JoinDate",
            when(col("JoinDate")=="blank","01-01-2023").otherwise(col("JoinDate"))
        )
    
    
df.show()
```

```python
df=df.replace('india',"India")
df=df.replace('New York',"USA")
df.show()
```

```python
df=df.replace("250","-200")
```

```python
df.display()
```

```python
df=df.withColumn("Sales",col("Sales").cast("int"))
df=df.filter(col("Sales")>=0)
df.display()
```
