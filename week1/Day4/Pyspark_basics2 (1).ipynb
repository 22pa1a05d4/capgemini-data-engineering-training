```python
from pyspark.sql.functions import col,when,try_to_date,date_format,current_date
df=spark.read.csv("/Volumes/workspace/default/sample/sample_dirty_data.csv",header=True,inferSchema=True)
df.display()
```

```python
df=df.fillna('blank')
df.display()
```

```python
df=df.withColumn(
    "Name",
    when(col("Name")=="blank","Unknown").otherwise(col("Name"))   
).withColumn(
    "City",
    when(col("City")=='blank','Unknown').otherwise(col("City"))
)
df.display()
```

```python
df=df.replace("Thirty","30")

df.display()
```

```python
df=df.replace('blank','0',subset=['Age','Sales'])

df.display()
```

```python
df=df.withColumn("Age",col("Age").cast("int"))
df.display()
```

```python
df=df.filter(col("Sales")>=0)
df.display()
```

```python
df=df.withColumn(
    "JoinDate",
    when(col("JoinDate")=="blank","01-01-2023").otherwise(col("JoinDate"))
)
df.display()
```

```python
df=df.withColumn(
    "JoinDate",
    try_to_date(col("JoinDate"),"dd-MM-yyyy"))
```

```python
df.display()
```

```python
df = df.fillna({"JoinDate": "2022-01-01"})
df.display()
```

```python
df=df.withColumn(
    "JoinDate",
    when(col("JoinDate").isNull(),try_to_date(current_date(),'dd-MM-yyyy')).otherwise(col("JoinDate"))
)
df.display()
```

```python
df=df.withColumn(
    "JoinDate",
    try_to_date(col("JoinDate"),"dd-MM-yyyy"))
df.display()
```

```python
df=df.dropDuplicates()
df.display()
```
