```python

revenue11=df.groupBy('dealer_city').agg(sum(col('price')*col('quantity')).alias('revenue')).orderBy(col("revenue").desc())
revenue11.display()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.window import Window
cust_rev=df.groupBy('cust_city','cust_name')\
    .agg(sum(col('price') * col('quantity')).alias('revenue'))
win=Window.partitionBy('cust_city').orderBy(col('revenue').desc())
top=cust_rev.withColumn('rank',rank().over(win)).filter(col('rank')<=3)
top.display()
```

```python
from pyspark.sql.functions import *
monthly_trend=df.withColumn('total',
                            col('price')*col('quantity'))\
    .withColumn('year',year(col('sale_date'))).withColumn('month',month(col('sale_date'))).groupBy('year','month').agg(sum('total').alias('revenue'))
monthly_trend.display()
```
