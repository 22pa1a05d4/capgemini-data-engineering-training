Basic GROUP BY Queries

```sql
select SUM(salary),department from Employee
group by department
```

```sql
select count(emp_id) AS Total_employees,department from Employee
group by department
```

```sql
select avg(salary),department from Employee
group by department
```

```sql
select MAX(salary),department from Employee
group by department 
```

```sql
select MIN(salary),department from Employee
group by department 
```


GROUP BY with Conditions


```sql
select SUM(salary) AS total,department from Employee
group by department
having total >100000
```

```sql
select count(emp_id) as Total_employees,department from Employee
group by department
having Total_employees > 2
```

```sql
select avg(salary), department from Employee
where joining_date>'2022-01-01'
group by department
```

```sql
select max(salary)as high,department from Employee
group by department
having high>75000
```

```sql
select sum(salary) as total,department from employee
group by department
having total<150000
```

GROUP BY with HAVING

```sql
select count(emp_id) as total_emp,department from employee
group by department
having total_emp>1
```

```sql
select sum(salary) as total_salary,department from employee
group by department
having total_salary>125000
```

```sql
select count(emp_id) as total_emp,department from employee
group by department
having total_emp>2
```

```sql
select avg(salary) as avg,department from employee
group by department
having avg>60000
```

```sql
select sum(salary),department from employee
group by department
having sum(salary) >=100000 and sum(salary)<=200000
```

GROUP BY with JOINs

```sql
select sum(amount),employee.emp_id from sales
join employee
on sales.emp_id = employee.emp_id
group by employee.emp_id
```

```sql
select count(sales_id),employee.emp_id from sales
inner join employee
on sales.emp_id=employee.emp_id
group by employee.emp_id
```

```sql
select sum(amount) as total_amount,sales.product from sales
inner join employee
on sales.emp_id=employee.emp_id
group by sales.product
```

```sql
select avg(amount),sales.product from sales inner join employee
on sales.emp_id=employee.emp_id
group by sales.product
```

```sql
select count(sales.amount) as sales_count,employee.emp_name from sales
inner join employee
on sales.emp_id=employee.emp_id
group by employee.emp_name
having sales_count>2
```

Advanced GROUP BY

```sql
select sum(employee.salary) as total_salary,sum(sales.amount) as total_amount,employee.emp_id from sales inner join employee on  sales.emp_id=employee.emp_id
group by employee.emp_id
```

```sql
select count(distinct(sales.product)),employee.emp_id from sales inner join employee on sales.emp_id=employee.emp_id
group by employee.emp_id
```

```sql
select max(sales.amount),employee.emp_id from sales inner join employee on sales.emp_id=employee.emp_id
group by employee.emp_id
```

```sql
select sum(amount) as total_sales,product from sales
group by product
having total_sales>50000
```

```sql
select avg(sales.amount),employee.department from sales inner join employee on sales.emp_id=employee.emp_id
group by employee.department
order by avg(sales.amount) desc
limit 1
```
