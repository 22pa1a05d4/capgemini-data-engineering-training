```sql
CREATE TABLE employee_payments (

emp_id INT PRIMARY KEY,

emp_name VARCHAR(50),

department VARCHAR(30),

base_salary DECIMAL(10,2),

bonus DECIMAL(10,2),

joining_date DATE

);
```

```sql
select initcap(emp_name) from employee_payments 
```

```sql
select emp_name,base_salary+ ifnull(bonus,0) as total_salary from employee_payments
```

```sql
select initcap(emp_name) ,round(base_salary + ifnull(bonus,0)) as total_salary from employee_payments
```

```sql
select initcap(emp_name) ,year(joining_date) from employee_payments
```

```sql
select initcap(emp_name),round(datediff(current_date(),joining_date)/365) as experience,
case 
when datediff(current_date(),joining_date)/365 >7 then 'Senior'
when datediff(current_date(),joining_date)/365 between 4 and 7 then 'Mid'
else 'Junior'
end as experience_category from employee_payments
```

Product Discount Validation


```sql
CREATE TABLE product_sales (

product_id INT,

product_name VARCHAR(50),

mrp DECIMAL(10,2),

selling_price DECIMAL(10,2),

sale_date DATE

);
```

```sql
INSERT INTO product_sales VALUES

(1,'Laptop',75000.75,68000.50,'2025-01-10'),

(2,'Mobile',35000.40,33000.25,'2025-01-12'),

(3,'Tablet',25000.90,26000.75,'2025-01-15');
```

```sql
select *,
case 
when selling_price<=mrp
then round((mrp-selling_price))
else 0
end as discount from product_sales
```

```sql
select *,
case 
when selling_price<=mrp
then round((mrp-selling_price)*100.0 /mrp)
else 0
end as discount_percent from product_sales
```

```sql
select dayname(sale_date) from product_sales
```

```sql
select *,
case 
when selling_price< mrp then  'Valid Discount'
when selling_price > mrp then 'over priced'
else 'No discount'
end as discount_status from product_sales
```
