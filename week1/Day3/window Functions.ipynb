ROW_NUMBER() 

```sql
select emp_id, emp_name, salary,
row_number() over(order by salary desc)
from employe
```

```sql
select emp_id,salary,department,
row_number() over(partition by  department order by salary desc) as rn
from employe
```

```sql
select row_number() over(order by join_date asc)as rn,*
from employe
```

```sql
select row_number() over(partition by department order by join_date)as rn,*
from employe
```

```sql
select row_number() over(order by salary asc) as rn,*
from employe
```

```sql
select row_number() over(order by emp_name asc) as rn,*
from employe
```

RANK()

```sql
select rank()  over(partition by department order by salary desc)as rn,*
from employe
```

```sql
select rank() over(partition by department order by salary asc)as rn,*
from employe
```

```sql
select rank() over(order by emp_name asc)as rn,*
from employe
```
