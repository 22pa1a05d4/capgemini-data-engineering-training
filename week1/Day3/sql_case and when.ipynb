```sql
select  emp_name, salary, experience, performance_rating,
CASE
when  performance_rating='A' and experience>=8 then salary *0.20
when performance_rating='B' and experience>=5 then salary *0.15
when performance_rating='C' then 0
else salary *0.10
end as hike
from Emp
```

```sql
select emp_name, department, salary, performance_rating,
case
when department='Finance' then 
case 
when performance_rating ='A' then salary*0.20
when performance_rating ='B' then salary*0.15
when performance_rating ='C' then salary*0.05
end 
when department='Engineering' then 
case
when performance_rating ='A' then salary*0.18
when performance_rating ='B' then salary*0.12
when performance_rating ='C' then salary*0.03
end
else salary*0.10
end as bonus
from emp
```

```sql
select emp_name, salary, performance_rating,
case 
when performance_rating ='A'  and salary >80000 then 'High Performer'
when performance_rating ='B'  and salary between 50000 and 80000 then 'Mid Performer'
when performance_rating ='C'  or salary <50000 then 'Low Performer'
else 'Average performer'
end as category
from emp
```

```sql
select emp_name, department, experience,
case 
when department='HR' then 
case 
when experience <5 then 'High risk'
when experience >=5 then 'Low risk'
end
when department='Finance'or department ='Engineering' then 
case 
when experience >8 then 'Low risk'
when experience <=8 then 'Medium risk'
end
else 'Medium Risk'
end as risk
from emp
```

```sql
select emp_name,performance_rating, department, salary, experience,
case 
when performance_rating ='A' then
case 
when salary >80000 and experience >5 then salary*0.25
when salary between 50000 and 80000 then salary*0.15
else salary*0.20
end
when performance_rating ='B' then
case 
when experience >5 then salary*0.12
else salary*0.10
end
when performance_rating ='C' then
case 
when salary <50000 then 0
else  0
end
else 0
end as hike
from emp

```

```sql
select emp_name,department,performance_rating,experience,
case 
when  department='Finance' then 
case 
when performance_rating ='A' and experience> 10 then salary*0.25
else salary*0.20
end
when department='HR' then 
case 
when performance_rating ='B' or experience> 5 then salary*0.15
else salary*0.10
end
else salary*0.08
end as bonus
from emp
```

```sql
select emp_name, salary, experience, performance_rating,
CASE
when salary >70000 then
case
when performance_rating='A' and experience>=8 then 'Top performer'
when experience<8 then 'Mid performer'
end
when salary between 50000 and 70000 then
case
when performance_rating='A' then 'Rising star'
else 'Average performer'
end
when salary <50000 then 'Low Performer'
end as performance
from emp
```

```sql
select emp_name, salary, experience,
case
when salary >90000 then
case
when experience> 10 then salary*0.35
else salary*0.30
end
when salary between 60000 and 90000 then
case 
when experience >5 then salary*0.25
else salary*0.20
end
when salary <60000 then salary*0.15
end as tax_rate
from emp
```

```sql
select emp_name, salary, experience,
case
when performance_rating ='A' then
case 
when salary >75000 and experience >7 then 'Eligible for Senior Role'
else 'Eligible for Junior Role'
end 
when performance_rating ='B' then
case 
when  experience >5 then 'Eligible for Consideration'
else 'Not Eligible'
end
when performance_rating ='C' then 'Not Eligible'
else 'Not Eligible'
end as eligibility
from emp
```
