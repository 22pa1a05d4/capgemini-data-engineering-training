```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col,upper,lower,trim
```

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("master_raw") \
    .getOrCreate()

df1 = spark.read.csv(
    "/Volumes/workspace/default/sample/Capgemini Databricks VITB-2026 (1).csv",
    header=True,
    inferSchema=True
)

df1.display()
```

```python
spark=SparkSession.builder\
    .appName("Task1_Responses")\
    .getOrCreate()
df2 = spark.read.csv("/Volumes/workspace/default/sample/Task 1 (Responses) (1).csv",header = True,inferSchema = True,multiLine=True)
df2.display()
```

```python
spark=SparkSession.builder\
    .appName("Task2_Responses")\
    .getOrCreate()
df3=spark.read.csv("/Volumes/workspace/default/sample/Task 1 file 2.csv",header=True,inferSchema=True, multiLine=True
    )
df3.display()
```

```python
df1 = df1.toDF(*[c.strip().lower().replace(" ", "_") for c in df1.columns])
df2 = df2.toDF(*[c.strip().lower().replace(" ", "_") for c in df2.columns])
df3=df3.toDF(*[c.strip().lower().replace(" ","_") for c in df3.columns])
```

Normalize emails 

```python
df1=df1.withColumn('college_domain_mailid',
                   lower(trim(col('college_domain_mailid')))
)
df1=df1.withColumn('personal_mail_id',
                   lower(trim(col('personal_mail_id')))
)
df1.display()
```

```python
df2=df2.withColumn('email_address',
                  lower(trim(col('email_address'))))
df2.display()
```

```python
df3=df3.withColumn('email_address',
                  lower(trim(col('email_address'))))
df3.display()
```

```python
from pyspark.sql.functions import array,explode
df1=df1.select(
    col("`s.no`"),col('superset_id'),col('regno'),col('student_name'),col('gender'),col('branch'),col('tenth_percentage'),col('inter_percentage'),col("`b.tech_cgpa`"),col('backlogs'),col('mobile'),col('college_domain_mailid'),col('personal_mail_id'),
    explode(array('college_domain_mailid','personal_mail_id')).alias('both_email')
)
df1.display()
```

```python
print(df1.columns)
```

```python
from pyspark.sql.functions import count
r1=df1.join(df2,df1.college_domain_mailid==df2.email_address,'inner')
r2=r1.filter(col('email_address').isNull())
r2.display()
```

```python
from pyspark.sql.functions import count,when,max
result1=df1.join(df2,df1.both_email==df2.email_address,'left')
result1=result1.withColumn('flag',
    when(col('email_address').isNotNull(),1).otherwise(0)) #created a column flag that gives weather the email is not present in table2 or not
result1=result1.groupBy('regno').agg(max('flag').alias('is_submitted')) # assign max value if both mails is 0 then not submitted
result1=result1.filter(col('is_submitted')==0) # shows values who dont submit
result1.display()
```

```python
new_inner=df1.join(df2,df1.both_email==df2.email_address,'inner')
new_inner=df1.join(df3,df1.both_email==df3.email_address,'inner')
new_inner.display()
```

```python
invalid=df2.join(df1,df1.both_email==df2.email_address,'left_anti') # left_anti gives all from table2 which is not present in table1
invalid.display()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number

w=Window.partitionBy("regno").orderBy("`s.no`") #  partition by regno
df1=df1.withColumn('rn',row_number().over(w)) # assigns row number
df1=df1.filter(col('rn')>1) # filter rows which has duplicates

df1.display()
```
