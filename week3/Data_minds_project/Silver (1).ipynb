```python
from pyspark.sql.functions import *
from pyspark.sql import SparkSession
spark=SparkSession.builder.getOrCreate()
```

```python
silver_encounters=spark.read.format('delta').load("/Volumes/data_minds/bronze/raw_data/Bronze_encounters")
silver_organizations=spark.read.format('delta').load("/Volumes/data_minds/bronze/raw_data/Bronze_organizations")
silver_patients=spark.read.format('delta').load("/Volumes/data_minds/bronze/raw_data/Bronze_patients")
silver_payers=spark.read.format('delta').load("/Volumes/data_minds/bronze/raw_data/Bronze_payers")
silver_procedures=spark.read.format('delta').load("/Volumes/data_minds/bronze/raw_data/Bronze_procedures")
```

```python
silver_encounters = silver_encounters.dropDuplicates(["Id"])
display(silver_encounters)
```

```python
silver_patients=silver_patients.dropDuplicates(['Id'])
display(silver_patients)
```

```python

silver_payers = silver_payers.dropDuplicates(["Id"])
display(silver_payers)
```

Encounters_Table_cleaning


```python
silver_encounters = silver_encounters.withColumnRenamed("id", "encounter_id") \
                       .withColumnRenamed("start", "start_time") \
                       .withColumnRenamed("stop", "end_time") \
                       .withColumnRenamed("patient", "patient_id") \
                       .withColumnRenamed("organization", "organization_id") \
                       .withColumnRenamed("payer", "payer_id") \
                       .withColumnRenamed("encounterclass", "encounter_class")\
                       .withColumnRenamed("BASE_ENCOUNTER_COST",'base_encounter_cost')\
                       .withColumnRenamed("CODE",'code')\
                           .withColumnRenamed("DESCRIPTION",'description')\
                           .withColumnRenamed("TOTAL_CLAIM_COST",'total_claim_cost')\
                           .withColumnRenamed("PAYER_COVERAGE",'payer_coverage')\
                           .withColumnRenamed("REASONCODE",'reason_code')\
                           .withColumnRenamed("REASONDESCRIPTION",'reason_description')
```

```python
silver_encounters = silver_encounters.withColumn("start_time", col("start_time").cast("timestamp")) \
                       .withColumn("end_time", col("end_time").cast("timestamp"))
```

```python
silver_encounters = silver_encounters.dropna(subset=[
    "encounter_id", "patient_id", "start_time"
])
```

```python
silver_encounters = silver_encounters.fillna({
    "reason_code": -1,
    "reason_description": "unknown"
})
```

```python
display(silver_encounters)
```

Patients_Silver_Transformations

```python
silver_patients = silver_patients.withColumnRenamed("id", "patient_id") \
    .withColumnRenamed('BIRTHDATE', "birth_date") \
    .withColumnRenamed("deathdate", "death_date") \
    .withColumnRenamed("prefix", "prefix") \
    .withColumnRenamed("suffix", "suffix") \
    .withColumnRenamed("maiden", "maiden") \
    .withColumnRenamed("marital", "marital") \
    .withColumnRenamed("race", "race") \
        .withColumnRenamed("ethnicity", "ethnicity") \
    .withColumnRenamed("gender", "gender") \
        .withColumnRenamed("birthplace", "birthplace") \
    .withColumnRenamed("address", "address") \
    .withColumnRenamed("city", "city") \
    .withColumnRenamed("state", "state") \
        .withColumnRenamed("county", "county") \
    .withColumnRenamed("zip", "zip") \
    .withColumnRenamed("lat", "lat") \
    .withColumnRenamed("lon","lon")\
                   .withColumnRenamed("first", "first_name") \
                   .withColumnRenamed("last", "last_name")
```

```python
silver_patients = silver_patients.dropna(subset=["patient_id", "birth_date"])
```

```python
silver_patients = silver_patients.fillna({
    "prefix": "unknown",
    "suffix": "unknown",
    "maiden": "unknown",
    "marital": "unknown",
    "race": "unknown",
    "ethnicity": "unknown",
    "gender": "unknown"
})
```

```python
silver_patients=silver_patients.withColumn(
    'marital',
    when(col('marital') == 'M', 'married')
    .when(col('marital') == 'S', 'single')
    .otherwise('unknown')
)
```

```python
silver_patients = silver_patients.withColumn("birthplace", split(col("birthplace"), "  "))

silver_patients = silver_patients.withColumn("birth_city", col("birthplace")[0]) \
                   .withColumn("birth_state", col("birthplace")[1]) \
                   .withColumn("birth_country", col("birthplace")[2])
```

```python
display(silver_patients)
```

Payers_Transformations

```python
silver_payers = silver_payers.withColumnRenamed("id", "payer_id") \
    .withColumnRenamed("name", "payer_name") \
    .withColumnRenamed("address", "payer_address") \
    .withColumnRenamed("city", "payer_city") \
    .withColumnRenamed("zip", "payer_zip") \
    .withColumnRenamed("phone", "payer_phone") \
               .withColumnRenamed("state_headquartered", "payer_state")
```

```python
silver_payers = silver_payers.fillna({
    "payer_address": "unknown",
    "payer_city": "unknown",
    "payer_state": "unknown",
    "payer_phone": "unknown"
})
```

```python

silver_payers = silver_payers.fillna({"payer_zip": -1})
```

```python
display(silver_payers)
```

Procedures_tranformations

```python
silver_procedures = silver_procedures.withColumnRenamed("start", "start_time") \
                       .withColumnRenamed("stop", "end_time") \
                       .withColumnRenamed("patient", "patient_id") \
                       .withColumnRenamed("encounter", "encounter_id") \
                       .withColumnRenamed("code", "reason_code") \
                       .withColumnRenamed("description", "reason_description")\
                       .withColumnRenamed('base_cost', 'base_procedure_cost')\
                           .withColumnRenamed('reasoncode', 'procedure_code')\
                            .withColumnRenamed('reasondescription', 'procedure_description')
```

```python
silver_procedures = silver_procedures.fillna({
    "reason_description": "unknown",
    "procedure_description": "unknown"
})
```

```python
silver_procedures = silver_procedures.dropna(subset=[
    "patient_id", "encounter_id", "start_time"
])
```

```python
display(silver_procedures)
```

Saving_data_to_schema

```python
silver_encounters.write.format('delta').mode('overwrite').save("/Volumes/data_minds/silver/preprocessed_data/silver_encounters")
silver_patients.write.format('delta').mode('overwrite').save("/Volumes/data_minds/silver/preprocessed_data/silver_patients")
silver_payers.write.format('delta').mode('overwrite').save("/Volumes/data_minds/silver/preprocessed_data/silver_payers")
silver_procedures.write.format('delta').mode('overwrite').save("/Volumes/data_minds/silver/preprocessed_data/silver_procedures")
```

```python
procedures_agg = silver_procedures.groupBy("encounter_id").agg(
    sum("base_procedure_cost").alias("total_procedure_cost"),
    count("*").alias("procedure_count")
)
```

```python
fact_df = silver_encounters \
    .join(silver_patients, on="patient_id", how="inner") \
    .join(silver_payers, on="payer_id", how="inner") \
    .join(procedures_agg, on="encounter_id", how="inner")
```

```python
fact_df.write.format('delta').mode('overwrite').save('/Volumes/data_minds/silver/preprocessed_data/fact_df')
```
