```python
gold_fact_df = gold_fact_df.withColumn(
    "full_name",
    concat_ws(
        " ",
        col("prefix"),
        col("first_name"),
        col("last_name"),
        when(col("suffix") == "unknown", None).otherwise(col("suffix"))
    )
)
```

## Innovative Healthcare KPIs

Comprehensive analysis of patient outcomes, financial performance, and operational efficiency

```python
# Calculate key financial metrics per encounter
kpi_df = gold_fact_df.withColumn(
    "encounter_duration_hours",
    (unix_timestamp("end_time") - unix_timestamp("start_time")) / 3600
).withColumn(
    "cost_recovery_rate",
    (col("payer_coverage") / col("total_claim_cost") * 100)
).withColumn(
    "out_of_pocket_cost",
    col("total_claim_cost") - col("payer_coverage")
).withColumn(
    "cost_per_hour",
    when(col("encounter_duration_hours") > 0, 
         col("total_claim_cost") / col("encounter_duration_hours")).otherwise(0)
).withColumn(
    "procedure_intensity_score",
    col("procedure_count") / col("encounter_duration_hours")
)
```

```python
# Segment patients by age and analyze costs
age_segments = kpi_df.withColumn(
    "age_group",
    when(col("patient_age") < 18, "0-17: Pediatric")
    .when(col("patient_age") < 35, "18-34: Young Adult")
    .when(col("patient_age") < 50, "35-49: Adult")
    .when(col("patient_age") < 65, "50-64: Middle Age")
    .when(col("patient_age") < 80, "65-79: Senior")
    .otherwise("80+: Elderly")
).groupBy("age_group").agg(
    count("encounter_id").alias("total_encounters"),
    countDistinct("patient_id").alias("unique_patients"),
    avg("total_claim_cost").alias("avg_cost_per_encounter"),
    sum("total_claim_cost").alias("total_revenue"),
    avg("procedure_count").alias("avg_procedures"),
    avg("encounter_duration_hours").alias("avg_duration_hours"),
    sum(when(col("death_status") == "deceased", 1).otherwise(0)).alias("deceased_count")
).withColumn(
    "mortality_rate_pct",
    (col("deceased_count") / col("unique_patients") * 100)
).withColumn(
    "cost_per_patient",
    col("total_revenue") / col("unique_patients")
)

display(age_segments.orderBy("age_group"))
```
